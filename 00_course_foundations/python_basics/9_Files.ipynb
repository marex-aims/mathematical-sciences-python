## Part 2. Opening and reading a file

To use a file we first **open** it: `open(name)` gives a **file handle**, a connection to the file. The handle is **not** the content.

**Notice:** The file handle is the way Python accesses the file; the content is the actual text or data stored inside it.


````python
handle = open('mails.txt')
print(handle)
````

The hanlde method `read()` returns the whole content as **one string**.

````python
text = handle.read()
print(len(text))
print(text[:100])
````

**Predict:** what does a second `handle.read()` return?

````python
print(len(handle.read()))
````

It returns an empty string: the handle has **already reached the end**. A handle is read **once**.

We close it when we are done:

````python
handle.close()
````

### The `with` statement

Forgetting `close()` is easy. The standard pattern closes the file for us at the end of the indented block:

````python
with open('mails.txt') as handle:
    text = handle.read()

print(len(text))
print(text[:40])
````

We can also print the whole content.

````python
print(text)
````

**Key idea:** we will always write `with open(name) as handle:`.

`text` is an ordinary string: everything you know about strings applies (slices, `len`, `split`...). The file is closed, but `text` stays available.

## Part 3. Reading line by line

A text file is a sequence of **lines**. Each line ends with the invisible **newline character** `\n`.

````python
print('Hello\nWorld')
print(repr('Hello\n'))       # repr shows the hidden characters
````

A handle can be used directly in a `for` loop: one line per turn.

**Live coding**: count the lines.

````python
count = 0
with open('mails.txt') as handle:
    for line in handle:
        count = count + 1

print('Number of lines:', count)
````

Now print the lines.

````python
with open('mails.txt') as handle:
    for line in handle:
        print(line)
````

Why the blank lines? Each `line` already ends with `\n`, and `print` adds another one. **`strip()`** removes the whitespace (spaces and `\n`) at both ends of a string.

````python
with open('mails.txt') as handle:
    for line in handle:
        print(line.strip())
````

## Part 4. Searching a file

Reading a file line by line and testing each line with a string method gives a **search**.

**Example**: print the subject lines.

````python
with open('mails.txt') as handle:
    for line in handle:
        line = line.strip()
        if line.startswith('Subject:'):
            print(line)
````

We can **count** the matching lines:

````python
count = 0
with open('mails.txt') as handle:
    for line in handle:
        if line.startswith('From:'):
            count = count + 1

print(count, 'senders')
````

`in` tests whether a text appears **anywhere** in the line. Notice that it also finds the `To:` line of the second message: `startswith` is more precise when we want one kind of line.

````python
with open('mails.txt') as handle:
    for line in handle:
        if 'ucad.sn' in line:
            print(line.strip())
````

### Extracting information

A line `From: awa@aims-senegal.org` has two parts separated by space ` `. `split()` gives a list, and the address is the element `[1]`. Splitting the address on `'@'` gives the user and the domain.

````python
line = 'From: awa@aims-senegal.org'
words = line.split()
print(words)

address = words[1]
user, domain = address.split('@')
print(user)
print(domain)
````

Now we build a **list of all the senders** from the file:

````python
senders = []
with open('mails.txt') as handle:
    for line in handle:
        if line.startswith('From:'):
            senders.append(line.split()[1])

print(senders)
````

**Warning:** test `startswith('From:')` with the colon. A message text could contain the word `From` elsewhere.

## Part 5. Files and dictionaries: who writes to us?

We now combine the file with the counting pattern you know. How many messages did each sender send?

````python
counts = {}
with open('mails.txt') as handle:
    for line in handle:
        if line.startswith('From:'):
            email_adresses = line.split()
            sender = email_adresses[1]
            counts[sender] = counts.get(sender, 0) + 1

print(counts)
````

## Part 6. Reading numbers

Everything read from a file is a **string**. To compute with numbers we must convert: `float(...)` or `int(...)`.

**Example**: obtaining the mean of the temperatures in `measures.txt` (one value per line).

````python
total = 0
count = 0
with open('measures.txt') as handle:
    for line in handle:
        value = float(line)        # float ignores the final newline
        total = total + value
        count = count + 1

print('total:', total)
print('count:', count)
print('mean:', total / count)
````

**Warning:** if the file has an **empty line**, `float('')` gives a `ValueError`. We skip empty lines with a test:

````python
values = []
with open('measures.txt') as handle:
    for line in handle:
        line = line.strip()
        if line != '':
            values.append(float(line))

print(values)
print(min(values), max(values))
````

`FileNotFoundError`: the program stops. Often a typo in the name is the cause. With **`try` / `except`** we try something risky and say what to do if it fails:


````python
try:
    handle = open('blabla.txt')
    print('file found')
except FileNotFoundError:
    print('There is no file with this name')
````

A useful function: count the lines of any file, and warn politely when the file is missing.

````python
def count_lines(filename):
    try:
        with open(filename) as handle:
            count = 0
            for line in handle:
                count = count + 1
            return count
    except FileNotFoundError:
        print('Sorry, I cannot open', filename)
        return None

print(count_lines('mails.txt'))
print(count_lines('blabla.txt'))
````

We can let the user choose the file with `input`:

````python
filename = input('File name: ')
n = count_lines(filename)
if n is not None:
    print(filename, 'has', n, 'lines')
````

## Part 8. Writing files

Open with the mode `'w'` (write). If the file does not exist, it is **created**. `write` does **not** add a newline: we add `'\n'` ourselves.


````python
with open('hello.txt', 'w') as handle:
    handle.write('AIMS Senegal\n')
    handle.write('Computing with Python\n')
````

Look at the file browser: `hello.txt` is there. Read it back:

````python
with open('hello.txt') as handle:
    print(handle.read())
````

**Warning:** the mode `'w'` **erases** an existing file. Run the next cell and look at the result:

````python
with open('hello.txt', 'w') as handle:
    handle.write('Only this line is left\n')

with open('hello.txt') as handle:
    print(handle.read())
````

To **add** at the end without erasing, use the mode `'a'` (append):

````python
with open('hello.txt', 'a') as handle:
    handle.write('A new line at the end\n')

with open('hello.txt') as handle:
    print(handle.read())
````

**`write` only accepts strings.** A number must be converted with `str`:

````python
with open('squares.txt', 'w') as handle:
    for n in range(1, 6):
        handle.write(str(n) + ' ' + str(n * n) + '\n')

with open('squares.txt') as handle:
    print(handle.read())
````

## Part 9. Summary and self-check

### Cheat sheet
```python
with open('data.txt') as handle:      # read (mode 'r' by default)
    text = handle.read()              # the whole file, one string

with open('data.txt') as handle:
    for line in handle:               # one line at a time
        line = line.strip()           # remove the newline

with open('out.txt', 'w') as out:     # write (erases!)
    out.write('text\n')              # we add the newline ourselves

with open('out.txt', 'a') as out:     # append (keeps the content)
    out.write('more\n')

try:
    handle = open(name)
except FileNotFoundError:
    print('no such file')
```

### Modes
| Mode | Meaning | If the file exists | If it does not exist |
|---|---|---|---|
| `'r'` | read | read it | `FileNotFoundError` |
| `'w'` | write | **erased** | created |
| `'a'` | append | content kept | created |

### Classic mistakes
| Mistake | Fix |
|---|---|
| `print(line)` gives a blank line after each line | `print(line.strip())` |
| `open('mails.txt')` fails: wrong folder or typo | check `os.getcwd()` and `os.listdir()` |
| reading the same handle twice gives `''` | store the result of `read()`, or open again |
| opening with `'w'` to add a line | use `'a'` |
| `handle.write(5)` gives a `TypeError` | `handle.write(str(5))` |
| `float('')` gives a `ValueError` | skip empty lines |
| the written text is on one line | add `'\n'` after each `write` |
