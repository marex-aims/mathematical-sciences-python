## The problem

Throw random points in a square with sides of length 1. A point `(x, y)` has `x` and `y` between 0 and 1.

The points with `x*x + y*y <= 1` are inside a quarter circle. The quarter circle has area $\pi / 4$. The square has area 1. So:

$$\pi \approx 4 \times \frac{\text{points inside}}{\text{all points}}$$

The same idea gives the area under a curve. The area under $y = x^k$ (for $x$ from 0 to 1) is exactly $\frac{1}{k+1}$. A point is under the curve if `y <= x**k`.

**Python hint:** `import random`, then `random.random()` gives a random number between 0 and 1.
