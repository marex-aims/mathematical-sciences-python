## The problem

We study a disease in a town with `N = 1000` people. Each person is in one group:

- **S**: can catch the disease
- **I**: has the disease
- **R**: had the disease and is now safe

Every day, the numbers change with these formulas. Use **today's** numbers for all of them:

- `new_infected = beta * S * I / N`
- `new_recovered = gamma * I`
- `S = S - new_infected`
- `I = I + new_infected - new_recovered`
- `R = R + new_recovered`

`beta` and `gamma` are numbers given to you. **Example** with `beta = 0.3`, `gamma = 0.1`, start `S = 999`, `I = 1`, `R = 0`:

- `new_infected = 0.3 * 999 * 1 / 1000 = 0.2997`
- `new_recovered = 0.1 * 1 = 0.1`
- day 1: `S = 998.7003`, `I = 1.1997`, `R = 0.1`

(We allow parts of a person. This is only a model.)

The number `R0 = beta / gamma` tells if the disease spreads. If `R0 > 1`, the number of sick people grows at first. If `R0 < 1`, it goes down.
