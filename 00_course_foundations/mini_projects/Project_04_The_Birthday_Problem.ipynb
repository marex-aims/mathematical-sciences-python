## The problem

How many people do you need in a room so that two of them probably have the same birthday? The answer is only **23**. This is surprising!

**Exact formula.** For `n` people and 365 days:

$$P(\text{all different}) = \frac{365}{365} \times \frac{364}{365} \times \frac{363}{365} \times \dots \text{ (n factors)}$$

$$P(\text{two people share a birthday}) = 1 - P(\text{all different})$$

**Example** for `n = 3`: P(all different) = (365/365) x (364/365) x (363/365) = 0.9918, so P(share) = 0.0082.

You will also check the formula with a **simulation**: give random birthdays to `n` people, many times, and count how often two people share a birthday.
