## The problem

A teacher has 20 exam scores. You will compute some numbers that describe the scores. Write the formulas yourself. Do **not** use the `statistics` module.

- **Mean:** sum of all scores divided by the number of scores.
- **Median:** sort the scores. If the number of scores is odd, take the middle one. If it is even, take the mean of the two middle ones.
- **Mode:** the score that appears most often.
- **Standard deviation** (how spread out the scores are):

$$\sigma = \sqrt{\frac{(x_1 - m)^2 + (x_2 - m)^2 + \dots + (x_n - m)^2}{n}}$$

where `m` is the mean. In Python, the square root of `x` is `x ** 0.5`.

- **Outlier:** a score with `abs(x - m) > 2 * sigma`.

**Example** for `[4, 8, 6, 5, 3, 8]`: mean 5.67, median 5.5, mode 8, standard deviation 1.89.
