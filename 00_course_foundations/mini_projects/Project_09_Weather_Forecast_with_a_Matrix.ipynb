## The problem

Tomorrow's weather depends only on today's weather. We write the chances in a **matrix** `P`. Row `i` gives the chances for tomorrow if today is state `i`.

Example with 2 states (Sunny, Rainy):

```python
P = [[0.8, 0.2],     # today Sunny: 80% Sunny tomorrow, 20% Rainy
     [0.4, 0.6]]     # today Rainy: 40% Sunny tomorrow, 60% Rainy
```

A **distribution** is a list of chances, like `[1, 0]` (today is Sunny for sure). To find tomorrow's distribution:

$$\text{new}[j] = \text{dist}[0] \cdot P[0][j] + \text{dist}[1] \cdot P[1][j] + \dots$$

**Example:** `dist = [1, 0]`.

- `new[0] = 1 * 0.8 + 0 * 0.4 = 0.8`
- `new[1] = 1 * 0.2 + 0 * 0.6 = 0.2`

So tomorrow is `[0.8, 0.2]`. The day after is `[0.72, 0.28]`.

After many days, the numbers stop changing. This is the **long-run** distribution.
