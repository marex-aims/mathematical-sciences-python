## The problem

Julius Caesar hid his messages: he replaced each letter by the letter `shift` places later in the alphabet. After `z` we go back to `a`.

Example with `shift = 3`: `hello` becomes `khoor`, and `xyz` becomes `abc`.

**Formula.** Give each letter a number: `a` = 0, `b` = 1, ..., `z` = 25.

$$\text{new number} = (\text{number} + \text{shift}) \bmod 26$$

**Python hint:**

- `ord(ch) - ord('a')` gives the number of the letter `ch`
- `chr(number + ord('a'))` gives the letter back
- `x % 26` is the remainder of `x` divided by 26 (so 25 + 3 gives 2)

To **break** the code: in English, the most common letter is `e`. So the most common letter in the secret message probably is the code for `e`.

In this project, all texts use only **lower-case** letters. Spaces and punctuation stay the same.
