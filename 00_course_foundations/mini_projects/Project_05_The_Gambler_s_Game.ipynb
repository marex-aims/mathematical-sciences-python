## The problem

A player has `start` coins. The player wants to reach `target` coins. In each round:

- with probability `p` the player **wins** 1 coin
- otherwise the player **loses** 1 coin

The game stops when the player has 0 coins (the player lost) or `target` coins (the player won).

**Formulas for a fair game** (`p = 0.5`):

$$P(\text{win}) = \frac{\text{start}}{\text{target}} \qquad \text{average number of rounds} = \text{start} \times (\text{target} - \text{start})$$

**Example:** `start = 5`, `target = 10`: P(win) = 0.5 and the average length is 5 x 5 = 25 rounds.

**Python hint:** `random.random() < p` is `True` with probability `p`.
